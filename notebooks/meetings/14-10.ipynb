## Findings Summary

### Overall
- **Spatial Judgment** — Correlation between *Placed Confidence* and *Distance*: **r = 0.25**, *p* < 0.0001  
- **Temporal Judgment** — Correlation between *Timed Confidence* and *Timed Correct*: **r = 0.07**, *p* = 0.0039  
- **Temporal Judgment** — Correlation between *Timed Confidence* and *Timed Error (days)*: **r = 0.09**, *p* = 0.0005  
- **Inter-Domain** — Correlation between *Placed Confidence* and *Timed Confidence*: **r = 0.61**, *p* < 0.0001  

### Per Participant
- **Spatial Judgment** — Mean within-participant correlation: **r = 0.213**, *t*(df) = 4.39, *p* = 1.0 × 10⁻⁴  
- **Temporal Judgment (Correct)** — Mean within-participant correlation: **r = 0.085**, *t*(df) = 1.86, *p* = 0.071  
- **Temporal Judgment (Error)** — Mean within-participant correlation: **r = 0.079**, *t*(df) = 1.86, *p* = 0.071

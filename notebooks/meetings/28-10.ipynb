## Questions regarding deuker et al. methods
### Comparing pearson values (an increase in pearson values, post-pre) to distances ?
- The brain has two ways of encoding information that are differentiable with pearson's coefficient -> positive or negative correlation with no information (random) with coefficient 0
- If we say that distance is positively correlated to pearson's values it means that when an image pair is more distant than another one it also has a higher pearson value. However This could mean two completely different things :
  Case 1 : We have two images very close with complete anti correlation -1 and two images very far with correlation 0 -> Images that are close together share information, far are completely random
  Case 2 : We have two images very close with complete anti correlation 0 and two images very far with correlation 1
  -> Images that are far from one another share information, close are completely random
-> linear scale of similarity debatable ?

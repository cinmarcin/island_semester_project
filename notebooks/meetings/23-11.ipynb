# Stimuli Effect

```python
fmri_path = 'data/processed/glm/per_trial/sub-P10/sub-P10_smooth_None_fitted_glm.pkl'
CONTRAST_TYPE = ContrastType.PER_TRIAL
fmri_glm = load(fmri_path)
contrast_vec = CONTRAST_TYPE.get_vector(fmri_glm.design_matrices_[0])
HEIGHT_CONTROL = 'bonferroni'
P_VALUE = 0.05
report = make_glm_report(fmri_glm, contrast_vec, height_control=HEIGHT_CONTROL, alpha=P_VALUE)
report
```
